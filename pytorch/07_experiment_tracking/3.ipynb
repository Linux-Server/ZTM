```python
# load the 07_effnetb2_data_20_percent_10_epochs
import torch
from torchvision.models import efficientnet_b2, EfficientNet_B2_Weights
from torch import nn

weights = torch.load("models/07_effnetb2_data_20_percent_10_epochs.pt")


model = efficientnet_b2()
model.classifier = nn.Sequential(
    nn.Dropout(p=0.3, inplace=True),
    nn.Linear(in_features=1408, out_features=3, bias=True)
    
)
model.load_state_dict(weights)

model


```

```python
# Now we need to prepare the input data format
from pathlib import Path
import requests
import os
url = "https://static.wixstatic.com/media/597497_39dfa709d3d845eeaf43eb692e93b31b~mv2.jpg/v1/fill/w_1500,h_944,al_c,q_85,usm_0.66_1.00_0.01,enc_avif,quality_auto/Image-empty-state.jpg"
```

```python
download_path = Path("download")

if not download_path.is_dir():
    download_path.mkdir()
    res = requests.get(url=url)
    image_path = download_path/"pizza_one.jpg"
    with open(image_path, "wb") as f:
        f.write(res.content)
    
    

```

```python
# get the automatic transformatiobn
automatic_transform = EfficientNet_B2_Weights.DEFAULT.transforms()

from torchvision.datasets import ImageFolder
from PIL import Image

image = Image.open(image_path)

```

```python
raw_image = automatic_transform(image)


raw_image.shape , raw_image.unsqueeze(dim=0).shape
```
